```python
import numpy as np
import matplotlib.pyplot as plt
from spatialmath import SE2, SE3
```

**Part A – Planar Mechanism (2D)**

Task 1 – Translation in XY Plane

```python
translations = [(2,3),(5,4),(3,-2),(-4,6)]
for i, (x,y) in enumerate(translations,1):
    T = SE2(x,y)
    print("Trial",i)
    print(T)
    print("----------------")
```

Task 2 – Rotation in 2D

```python
angles = [30,60,90,180]
for angle in angles:
    R = SE2.Rot(angle,"deg")
    print("Rotation:",angle,"degrees")
    print(R)
    print("----------------")
```

Task 3 – Translation Followed by Rotation

```python
cases = [((3,1),45),((5,2),60),((2,6),120)]
for i,(trans,angle) in enumerate(cases,1):
    T = SE2(trans[0],trans[1])
    R = SE2.Rot(angle,"deg")
    H = T*R
    print("Case",i)
    print("Translation:",trans)
    print("Rotation:",angle)
    print(H)
    print("----------------")
```

Task 4 – Compare Transformation Order

```python
T = SE2(4,2)
R = SE2.Rot(90,"deg")
TR = T*R
RT = R*T
print("Translation followed by Rotation")
print(TR)
print("\nRotation followed by Translation")
print(RT)
print("\nAre they equal?")
print(TR == RT)
```

**Part B – Spatial Mechanism (3D)**

Task 5 – Translation in XYZ Space

```python
translations_3D = [(4,2,1),(5,5,5),(-2,3,7)]
for i,t in enumerate(translations_3D,1):
    T = SE3(t[0],t[1],t[2])
    print("Trial",i)
    print(T)
    print("----------------")
```

Task 6 – Rotation About X-axis

```python
angles=[30,90,180]
for angle in angles:
    Rx = SE3.Rx(angle,"deg")
    print("X Rotation:",angle)
    print(Rx)
    print("----------------")
```

Task 7 – Rotation About Y-axis

```python
angles=[45,120,180]
for angle in angles:
    Ry = SE3.Ry(angle,"deg")
    print("Y Rotation:",angle)
    print(Ry)
    print("----------------")
```

Task 8 – Rotation About Z-axis

```python
angles=[60,90,270]
for angle in angles:
    Rz = SE3.Rz(angle,"deg")
    print("Z Rotation:",angle)
    print(Rz)
    print("----------------")
```

Task 9 – Combined Translation and Rotation

```python
cases = [((5,4,2),30,45,60),((2,8,3),45,30,90),((1,5,4),60,60,60)]
for i,(pos,x,y,z) in enumerate(cases,1):
    T = SE3(pos[0],pos[1],pos[2])
    Rx = SE3.Rx(x,"deg")
    Ry = SE3.Ry(y,"deg")
    Rz = SE3.Rz(z,"deg")
    H = T*Rx*Ry*Rz
    print("Case",i)
    print("Final Transformation Matrix")
    print(H)
    print("----------------")
```

Task 10 – Coordinate Frame Visualization

```python
A = SE3(2, 3, 4) * SE3.Rz(45, "deg")
B = SE3(4, 2, 1) * SE3.Rx(30, "deg")
C = SE3(1, 5, 2) * SE3.Ry(60, "deg")
A.plot(frame='A')
B.plot(frame='B')
C.plot(frame='C')
plt.title("Multiple Coordinate Frames")
plt.show()
```

**Challenge Exercise**

```python
T1 = SE3(3,2,1)
Rx = SE3.Rx(45,"deg")
Ry = SE3.Ry(60,"deg")
Rz = SE3.Rz(30,"deg")
T2 = SE3(1,2,0)
Final_T = T1 * Rx * Ry * Rz * T2
print("Final Homogeneous Transformation Matrix")
print(Final_T)
```

```python
position = Final_T.t
print("Final End Effector Position:")
print(position)
```

```python
Final_T.plot(frame='End')
plt.title("Final End Effector Frame")
plt.show()
```

```python
Alternative_T = T2 * Rz * Ry * Rx * T1
print("Original Transformation")
print(Final_T)
print("\nChanged Order Transformation")
print(Alternative_T)
print("\nAre both same?")
print(Final_T == Alternative_T)
```
